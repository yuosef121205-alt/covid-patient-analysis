```python
import pandas as pd

df = pd.read_csv("Covid Data.csv")

print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
display(df.head())
```

```python

# Check data types and missing values
df.info()

# Statistical summary of age
print("\nAge Summary:")
print(df["AGE"].describe())

# Count patients by patient type
print("\nPatient Type Counts:")
print(df["PATIENT_TYPE"].value_counts(dropna=False))

# Check the unique patient type codes
print("\nUnique Patient Types:")
print(df["PATIENT_TYPE"].unique())
```

```python

print("Minimum age:", df["AGE"].min())
print("Maximum age:", df["AGE"].max())
print("Average age:", df["AGE"].mean())
print("Median age:", df["AGE"].median())

print("\nPatient Type Counts:")
print(df["PATIENT_TYPE"].value_counts(dropna=False))

print("\nPatient Type Percentages:")
print(
    df["PATIENT_TYPE"].value_counts(normalize=True, dropna=False)
    .mul(100)
    .round(2)
)
```

```python

print("PATIENT_TYPE codes:")
print(df["PATIENT_TYPE"].value_counts().sort_index())

print("\nDATE_DIED sample values:")
print(df["DATE_DIED"].value_counts().head(10))

print("\nICU codes:")
print(df["ICU"].value_counts().sort_index())
```

```python

conditions = ["DIABETES", "HIPERTENSION", "OBESITY"]

for col in conditions:
    print(f"{col}:")
    print(df[col].value_counts().sort_index())
    print()
```

```python

hospitalized_count = (df["PATIENT_TYPE"] == 2).sum()
total_patients = len(df)

hospitalization_rate = hospitalized_count / total_patients * 100

print("Hospitalized Patients:", hospitalized_count)
print("Hospitalization Rate:", round(hospitalization_rate, 2), "%")
```

```python

hospitalized = df[df["PATIENT_TYPE"] == 2]
non_hospitalized = df[df["PATIENT_TYPE"] == 1]

print("Average age of hospitalized patients:",
      hospitalized["AGE"].mean())

print("Average age of non-hospitalized patients:",
      non_hospitalized["AGE"].mean())
```

```python

older_hospitalized = df[
    (df["AGE"] > 60) &
    (df["PATIENT_TYPE"] == 2)
]

print("Hospitalized patients older than 60:",
      len(older_hospitalized))
```

```python

diabetes_patients = df[df["DIABETES"] == 1]

hospitalized_diabetes = diabetes_patients[
    diabetes_patients["PATIENT_TYPE"] == 2
]

rate = len(hospitalized_diabetes) / len(diabetes_patients) * 100

print("Diabetes patients:", len(diabetes_patients))
print("Hospitalized diabetes patients:", len(hospitalized_diabetes))
print("Hospitalization rate:", round(rate, 2), "%")
```

```python

hypertension_patients = df[df["HIPERTENSION"] == 1]

hospitalized_hypertension = hypertension_patients[
    hypertension_patients["PATIENT_TYPE"] == 2
]

rate = len(hospitalized_hypertension) / len(hypertension_patients) * 100

print("Hypertension patients:", len(hypertension_patients))
print("Hospitalized hypertension patients:", len(hospitalized_hypertension))
print("Hospitalization rate:", round(rate, 2), "%")
```

```python

pneumonia_patients = df[df["PNEUMONIA"] == 1]

hospitalized_pneumonia = pneumonia_patients[
    pneumonia_patients["PATIENT_TYPE"] == 2
]

rate = len(hospitalized_pneumonia) / len(pneumonia_patients) * 100

print("Pneumonia patients:", len(pneumonia_patients))
print("Hospitalized pneumonia patients:", len(hospitalized_pneumonia))
print("Hospitalization rate:", round(rate, 2), "%")
```

```python
bins = [-1, 18, 30, 45, 60, 75, float("inf")]
labels = ["0-18", "19-30", "31-45", "46-60", "61-75", "76+"]

df["AGE_GROUP"] = pd.cut(
    df["AGE"],
    bins=bins,
    labels=labels
)

print(df[["AGE", "AGE_GROUP"]].head(10))
```

```python

print(df["AGE_GROUP"].value_counts().sort_index())
```

```python

hospitalized_by_age = (
    df[df["PATIENT_TYPE"] == 2]
    .groupby("AGE_GROUP", observed=False)
    .size()
)

print(hospitalized_by_age)
```

```python

hospitalization_rates = (
    df.assign(HOSPITALIZED=(df["PATIENT_TYPE"] == 2))
    .groupby("AGE_GROUP", observed=False)
    ["HOSPITALIZED"]
    .mean()
    .mul(100)
    .round(2)
)

print(hospitalization_rates)
```

```python
both_conditions = (
    (df["DIABETES"] == 1) &
    (df["HIPERTENSION"] == 1)
)

neither_condition = (
    (df["DIABETES"] == 2) &
    (df["HIPERTENSION"] == 2)
)

print("Both conditions:", both_conditions.sum())
print("Neither condition:", neither_condition.sum())
```

```python
conditions = [
    "DIABETES",
    "HIPERTENSION",
    "OBESITY",
    "ASTHMA",
    "COPD",
    "CARDIOVASCULAR"
]

df["CONDITION_COUNT"] = (df[conditions] == 1).sum(axis=1)

print(df[["CONDITION_COUNT"]].head(10))
```

```python
df["CONDITION_COUNT"].value_counts().sort_index()
```

```python
hospitalization_by_conditions = (
    df.assign(HOSPITALIZED=(df["PATIENT_TYPE"] == 2))
    .groupby("CONDITION_COUNT")["HOSPITALIZED"]
    .mean()
    .mul(100)
    .round(2)
)

print(hospitalization_by_conditions)
```

```python
risk_factors = ["DIABETES", "HIPERTENSION", "PNEUMONIA"]

for col in risk_factors:
    patients = df[df[col] == 1]
    rate = (patients["PATIENT_TYPE"] == 2).mean() * 100
    print(f"{col}: {rate:.2f}%")
```
